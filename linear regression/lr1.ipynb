## Tools
In this project, i made use of: 
- NumPy, a popular library for scientific computing
- Matplotlib, a popular library for plotting data

```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# These are the variables used to train the model
x_train = np.array([1, 2])
y_train = np.array([300, 500])
```

```python
m = x_train.shape[0]
print(f"Number of training examples: {m}")
```

```python
# getting the ith data
i = 0

x_i = x_train[i]
y_i = y_train[i]

print(f"x_{i}, y_{i}: {x_i}, {y_i}")
```

```python
# Plotting the data
plt.scatter(x_train, y_train, marker = "x")

# setting the title
plt.title("Housing prices")
plt.xlabel("Size (in sq ft)")
plt.ylabel("Price (in 1000s of dollars)")
plt.show()
```

```python
# setting the weight and bias
w = 120
b = 150
print(f"w : {w}")
print(f"b : {b}")
```

```python
def compute_model_output(x, w, b):

    """
    Computes the prediction of a linear model
    Args:
      x (ndarray (m,)): Data, m examples 
      w,b (scalar)    : model parameters  
    Returns
      f_wb (ndarray (m,)): model prediction

    """

    m = x_train.shape[0]
    f_wb = np.zeros(m)

    for i in range(m):
        f_wb[i] = w*x[i] + b

    return f_wb
```

```python
tmp_f_wb = compute_model_output(x_train, w, b)

# plot our model predictions
plt.plot(x_train, tmp_f_wb, c = "b", marker = "x", label = "Our predictions")

# plot the actual data
plt.plot(x_train, y_train, c = "r", marker = "x", label = "Actual data")

plt.title("Housing prices")
plt.xlabel("Size (in 1000s of sq ft)")
plt.ylabel("Price (in 1000s of dollars)")

plt.legend()
plt.show()
```

By training the model, we can use it to predict the price for the house using it's area.
