### Single prediction one by one using a loop
This is the demonstration of prediction before training the model

```python
def predict_simple_loop(x, w, b):
    """
    single predict using linear regression
        
        Args:
        x (ndarray): Shape (n,) example with multiple features
        w (ndarray): Shape (n,) model parameters    
        b (scalar):  model parameter     
        
        Returns:
        p (scalar):  prediction
    """

    p = 0
    for i in range(x.shape[0]):
        p_i = w[i]*x[i]
        p = p + p_i

    p = p + b

    return p
```

```python
# get a row from our training data
x = x_train[0, :]
print(f"X : {x}, X-Shape : {x.shape}")

price = predict_simple_loop(x, w_init, b_init)

print(f"Price before training the model: {price}" )
```

### Single prediction parallelly

```python
def single_prediction(x, w, b):
    """ 
    single predict using linear regression
        
        Args:
        x (ndarray): Shape (n,) example with multiple features
        w (ndarray): Shape (n,) model parameters    
        b (scalar):  model parameter     
        
        Returns:
        p (scalar):  prediction
    """
    p = np.dot(w, x) + b

    return p
```

```python
# get a row from our training data
x = x_train[0, :]
print(f"X : {x}, X-Shape : {x.shape}")

price = single_prediction(x, w_init, b_init)

print(f"Price before training the model using vectors: {price}" )
```

### Compute cost with multiple variables


```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    c = 0
    
    for i in range(m):
        j = (np.dot(w,x[i]) + b - y[i]) ** 2
        c = c + j
    
    c = (1/2*m) * c
    return c
```
